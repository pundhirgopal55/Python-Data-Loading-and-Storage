```python
#csv file using read and write
import pandas as pd

# Separate the header from the row data
data = [
    ["Ayaansh", 19, "Computer Science"],
    ["Ansh", 20, "Mechanical"],
    ["Anshika", 23, "Civil"],
    ["Rohan", 21, "Electronics"]
]
columns = ["Name", "Age", "Department"]
# Create the DataFrame and save it to a CSV file
df = pd.DataFrame(data, columns=columns)
df.to_csv("name.csv", index=False)
print(df)

df.to_json("name.json", orient="records", indent=4)
print(df.to_json(orient="records", indent=4))

df.to_excel("name.xlsx", index=False)
print(df)

df.to_csv("name.tsv", sep="\t", index=False)
print("--- Tab Separated Output ---")
print(df.to_csv(sep="\t", index=False))
fwf_data = df.to_string(index=False)
with open("name.txt", "w") as f:
    f.write(fwf_data)
    print(fwf_data)
```

# Sqlite

```python
import sqlite3
conn = sqlite3.connect("students.db")
```

```python
cursor =  conn.cursor()
print("DataBase Created Successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS student_table(
NAME TEXT,
ROLL_NO INTEGER,
MARKS INTEGER )
"""
)
print("Table is created successfully !!")
```

```python
#write the data into table :-
cursor.execute(
    "INSERT INTO student_table VALUES (? ,? ,?)",
    ("Amit",101,90)
)
cursor.execute(
    "INSERT INTO student_table VALUES ('Aman',102,95) "
)
```

```python
conn.commit()
print("Records inserted successfully !!")
conn.close()
```

```python
#Retrive | read the data
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM student_table")
rows = cursor.fetchall()
for row in rows :
  print(row)
  conn.close()
```
